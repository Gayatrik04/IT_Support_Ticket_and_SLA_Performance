```python
date_columns = [
    "Created time","Expected SLA to resolve","Expected SLA to first response","First response time","Resolution time","Close time"]

for col in date_columns:
    df[col] = pd.to_datetime(df[col])
```

```python
df[df["Agent interactions"].isnull()]["Status"].value_counts()
df["Agent interactions"] = df["Agent interactions"].fillna(0).astype(int)
```

```python
df["Status"].unique()
df["Priority"].unique()
df["Source"].unique()
df["Support Level"].unique()
df["Agent Group"].unique()
df["Product group"].unique()
df['Country'].unique()
df['Topic'].unique()
df['SLA For first response'].unique()
df['SLA For Resolution'].unique()
```

```python
df.describe()
```

```python
df["Year"] = df["Created time"].dt.year
df["Month"] = df["Created time"].dt.month_name()
df["Month Number"] = df["Created time"].dt.month
df["Day"] = df["Created time"].dt.day
df["Weekday"] = df["Created time"].dt.day_name()
df["Hour"] = df["Created time"].dt.hour
df["Day Type"] = df["Weekday"].apply(
    lambda x: "Weekend" if x in ["Saturday", "Sunday"] else "Weekday"
)
```

```python
df["Working Hours"] = df["Hour"].apply(
    lambda x: "Business Hours" if 9 <= x < 18 else "After Hours"
)

df["Response Time (Minutes)"] = (df["First response time"] - df["Created time"]).dt.total_seconds() / 60

df["Resolution Time (Hours)"] = (df["Resolution time"] - df["Created time"]).dt.total_seconds() / 3600

df["Close Duration (Hours)"] = (df["Close time"] - df["Resolution time"]).dt.total_seconds() / 3600

df["Response Delay (Minutes)"] = (df["First response time"] - df["Expected SLA to first response"]).dt.total_seconds() / 60
```

```python
df.columns = (df.columns.str.replace(" ", "_").str.replace("(", "", regex=False).str.replace(")", "", regex=False))
```

```python
df=df.to_csv('Technical_support_dataset.csv',index=False)
```
